### Step 1: Environment Setup & Data Ingestion

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Import data
data = pd.read_csv('/content/googleplaystore.csv')
data.head()
```

```python
# Check structural health and data types
data.info()
```

### Step 2: Data Transformation & Type Casting

**The Issue:**
Initial inspection reveals that core numerical columns (`Installs`, `Price`, `Size`) are read by Pandas as `object` (strings). This is due to the presence of special formatting characters (e.g., `+`, `,`, `$`, `M`, `k`) and inconsistent text entries (e.g., "Varies with device").

**The Action:**
To perform quantitative analysis for the SwiftFlow launch, we must systematically strip these non-numeric characters and cast the variables to proper numeric data types (integers and floats).

```python
data[data['Installs'] == 'Free']
```

```python
# Drop the corrupted row by its index number
data = data.drop(10472)
```

```python
# Transform Installs column
data['Installs'] = data['Installs'].str.replace('+','',regex = False)
data['Installs'] = data['Installs'].str.replace(',', '', regex= False)

# Convert the column to an integer
data['Installs'] = data['Installs'].astype(int)
```

```python
# Transform Price column
data['Price'] = data['Price'].str.replace('$','', regex = False)

# Convert the column to Numeric
data['Price'] = data['Price'].astype(float)
```

```python
data['Size'].unique()
```

```python
import numpy as np

def clean_size(size_val):
    if size_val == 'Varies with device':
        return np.nan
    elif size_val[-1] == 'M':
        return float(size_val[:-1]) # Already in Megabytes
    elif size_val[-1] == 'k':
        return float(size_val[:-1]) / 1024 # Convert KB to MB
    else:
        return np.nan # Catch any other weird edge cases
```

```python
# Apply the function to the Size column
data['Size'] = data['Size'].apply(clean_size)

# Rename the column
data.rename(columns={'Size': 'Size_MB'}, inplace=True)

data.info()
```

### Step 3: Duplicate Purge
Removing exact duplicate rows to prevent artificial inflation of popular apps in our downstream analysis.

```python
# Count exact duplicate rows
duplicate_count = data.duplicated().sum()
print(f"Total Duplicated Rows : {duplicate_count}")
```

```python
print("Original Shape :", data.shape)

# Drop Duplicate Rows
data = data.drop_duplicates()

print("Cleaned Shape:", data.shape)
```

### Step 4: Null Value Imputation
*   **Minor Categoricals:** Dropping rows with missing `Type`, `Current Ver`, and `Android Ver` (negligible data loss).
*   **Rating:** Imputing missing values using the **Mean** rating of the app's specific `Category`.
*   **Size_MB:** Imputing missing values using the **Median** size of the app's specific `Category` to avoid outlier skew.

```python
# Count null values in every column
print("\n--- Missing Values ---")
print(data.isnull().sum())
```

```python
# Drop minor categorical nulls
data.dropna(subset=['Type','Current Ver', 'Android Ver'], inplace= True)


# Impute Rating with Category Mean
data['Rating'] = data['Rating'].fillna(
    data.groupby('Category')['Rating'].transform('mean')
)


# Impute Size_MB with Category Median
data['Size_MB'] = data['Size_MB'].fillna(
    data.groupby('Category')['Size_MB'].transform('median')
)

# Verify
print("--- Missing Values After Imputation ---")
print(data.isnull().sum())
```

### Step 5: Feature Engineering (Genres)
The raw `Genres` column contains multiple sub-genres separated by semicolons (e.g., 'Puzzle;Brain Games'). To ensure accurate aggregation during EDA, we are engineering a `Primary_Genre` feature that isolates only the root category.

```python
data['Genres'].unique()[:20]
```

```python
# Create the new feature by splitting the string at the semicolon and taking the first item

data['Primary_Genre'] = data['Genres'].str.split(';').str[0]

print("----- OLD VS NEW -----")
print(data[['Genres', 'Primary_Genre']].head(15))
```

### Step 6: Temporal Data Conversion
Converting the `Last Updated` string column into a formal Pandas `datetime` object to allow for time-series analysis and cohort tracking.

```python
# Convert the string dates to datetime objects
data['Last Updated'] = pd.to_datetime(data['Last Updated'])

data.info()
```
